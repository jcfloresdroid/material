# Distribución de frecuencias (versión libro)

## ¿Qué es una distribución de frecuencias?

Imagínate que el profe te da las 80 calificaciones así, sueltas. Si te pregunta *"¿cómo le fue al grupo?"*, viendo la lista no puedes decir mucho, ¿verdad? Son puros números revueltos.

Una **distribución de frecuencias** es simplemente **acomodar los datos en grupos (clases o intervalos) y contar cuántos caen en cada grupo**. Es como cuando separas la ropa sucia por colores: ya no ves 80 prendas, ves "15 blancas, 30 de color, 35 oscuras" y entiendes todo de un vistazo.

Para armarla seguimos estos pasos:

1. Calcular el **rango** (qué tan separados están el dato más chico y el más grande).
2. Decidir **cuántas clases** queremos.
3. Calcular el **ancho (tamaño) de clase**.
4. Escribir los **límites de clase**.
5. Calcular los **límites reales / fronteras de clase**.
6. Calcular las **marcas de clase**.
7. **Contar** cuántos datos caen en cada clase (frecuencia) y sacar las frecuencias relativas y acumuladas.

## Paso 1: El rango

El rango ya lo calculamos en el inciso c), pero lo repetimos para tenerlo a la mano:

$$\text{Rango} = \text{dato mayor} - \text{dato menor}$$

```python
n = len(calificaciones)          # cuántos datos tenemos
dato_mayor = calificaciones.max()
dato_menor = calificaciones.min()
rango = dato_mayor - dato_menor

print("Número de datos (n):", n)
print("Dato mayor:", dato_mayor)
print("Dato menor:", dato_menor)
print("Rango:", rango)
```

## Paso 2: ¿Cuántas clases usamos?

No hay una regla única, pero lo normal es usar **entre 5 y 20 clases**. Si usas muy poquitas, todo se amontona y no ves nada; si usas demasiadas, quedan clases casi vacías y tampoco ves nada.

Una fórmula muy usada para darnos una idea es la **regla de Sturges**:

$$k = 1 + 3.322 \cdot \log_{10}(n)$$

Vamos a ver qué nos sugiere:

```python
k_sturges = 1 + 3.322 * math.log10(n)
print("Sturges sugiere:", round(k_sturges, 2), "clases")
```

Sturges nos dice más o menos **7 clases**. Pero ojo: es solo una sugerencia, no una ley.

El libro decide usar **10 clases de ancho 5**, porque así los intervalos quedan bonitos: 50-54, 55-59, 60-64... Eso es súper cómodo para leer calificaciones. Nosotros le vamos a hacer caso al libro.

## Paso 3: Ancho (tamaño) de clase

El ancho de clase se saca con:

$$c = \frac{\text{Rango}}{\text{número de clases}}$$

y normalmente **se redondea hacia arriba** a un número cómodo.

```python
k = 10                       # número de clases que elegimos (como el libro)
c_calculado = rango / k
print("Ancho calculado:", c_calculado)

c = 5                        # lo redondeamos a un número cómodo
print("Ancho de clase que usaremos (c):", c)
```

Salió 4.4, y lo redondeamos a **5**. Con 10 clases de ancho 5 cubrimos 50 valores (del 50 al 99), y nuestros datos van del 53 al 97, así que cabe todo sin problema. 👌

## Paso 4: Límites de clase

Los **límites de clase** son los números que "se ven" en cada intervalo. Por ejemplo, en la clase **50 - 54**:

- **Límite inferior:** 50
- **Límite superior:** 54

Empezamos en 50 (un número redondo un poquito abajo del dato menor, que es 53) y vamos sumando el ancho `c = 5`:

```python
# Límites inferiores: 50, 55, 60, ... , 95
limites_inf = list(range(50, 100, c))

# Límite superior = límite inferior + c - 1  (porque son números enteros)
limites_sup = [li + c - 1 for li in limites_inf]

for li, ls in zip(limites_inf, limites_sup):
    print(f"{li} - {ls}")
```

## Paso 5: Límites reales (fronteras de clase)

Aquí viene algo que a muchos nos confunde al principio, así que vamos despacito. 🐢

Mira las clases **50 - 54** y **55 - 59**. Entre el 54 y el 55 hay un "huequito". ¿Y si alguien hubiera sacado 54.7? ¿En qué clase lo pondríamos? 🤔

Para que no haya huecos, se usan los **límites reales**: se le **resta medio punto (0.5)** al límite inferior y se le **suma medio punto** al límite superior:

| Clase | Límites reales |
|-------|----------------|
| 50 - 54 | 49.5 - 54.5 |
| 55 - 59 | 54.5 - 59.5 |
| ... | ... |

¿Ves? Ahora el final de una clase (54.5) es **exactamente** el inicio de la siguiente (54.5). Ya no hay huecos, las clases quedan "pegaditas".

Esos puntos donde una clase termina y empieza la otra (49.5, 54.5, 59.5, ...) se llaman **fronteras de clase**. En la práctica, **límites reales y fronteras de clase son lo mismo**: la frontera de clase es el punto que comparten dos clases vecinas.

> 💡 ¿Por qué 0.5? Porque nuestros datos son enteros (van de 1 en 1), y la mitad de esa unidad es 0.5. Si los datos tuvieran un decimal (como 7.3, 7.4), restaríamos 0.05.

Con los límites reales, el ancho de clase se ve clarito: $54.5 - 49.5 = 5$ ✅

```python
limites_reales_inf = [li - 0.5 for li in limites_inf]
limites_reales_sup = [ls + 0.5 for ls in limites_sup]

for li, ls, lri, lrs in zip(limites_inf, limites_sup, limites_reales_inf, limites_reales_sup):
    print(f"Clase {li} - {ls}   ->   Límites reales {lri} - {lrs}")
```

Y las **fronteras de clase** son todos esos puntos de corte juntos, sin repetir (son 11 porque 10 clases necesitan 11 "paredes"):

```python
fronteras = limites_reales_inf + [limites_reales_sup[-1]]
print("Fronteras de clase:", fronteras)
```

## Paso 6: Marcas de clase

La **marca de clase** es el **punto medio** de cada clase. Es como el "representante" del intervalo: cuando ya agrupamos los datos, hacemos de cuenta que todos los de esa clase valen la marca de clase.

$$X_m = \frac{\text{límite inferior} + \text{límite superior}}{2}$$

Por ejemplo, para la clase 50 - 54: $\frac{50 + 54}{2} = 52$.

(Da lo mismo si lo haces con los límites reales: $\frac{49.5 + 54.5}{2} = 52$. 😉)

```python
marcas_clase = [(li + ls) / 2 for li, ls in zip(limites_inf, limites_sup)]
print("Marcas de clase:", marcas_clase)
```

## Paso 7: Contar las frecuencias

Ahora sí, a contar. Para cada clase vemos cuántas calificaciones caen entre su límite inferior y su límite superior. Esto es la **frecuencia absoluta** ($f$).

Lo hacemos con la misma idea que usamos en los incisos g) y h): ponemos una condición y la sumamos con `.sum()` (cada `True` cuenta como 1).

```python
frecuencias = []
for li, ls in zip(limites_inf, limites_sup):
    cuantos = ((calificaciones >= li) & (calificaciones <= ls)).sum()
    frecuencias.append(int(cuantos))   # int() para que se imprima como número normal

print("Frecuencias:", frecuencias)
print("Suma de frecuencias:", sum(frecuencias))   # debe dar 80
```

¡La suma da 80! Eso quiere decir que no se nos perdió ni se nos duplicó ningún alumno. Siempre haz esta revisión. ✅

Ahora agregamos más columnas útiles:

- **Frecuencia relativa ($f_r$)**: qué fracción del total cae en la clase. $f_r = \frac{f}{n}$
- **Porcentaje**: lo mismo pero multiplicado por 100.
- **Frecuencia acumulada ($F$)**: cuántos datos llevamos "juntados" hasta esa clase (vas sumando de arriba hacia abajo).
- **Frecuencia relativa acumulada ($F_r$)**: lo mismo pero en fracción del total.

Y lo metemos todo en un `DataFrame` de pandas para que se vea como tabla:

```python
tabla = pd.DataFrame({
    "Clase": [f"{li} - {ls}" for li, ls in zip(limites_inf, limites_sup)],
    "Límites reales": [f"{lri} - {lrs}" for lri, lrs in zip(limites_reales_inf, limites_reales_sup)],
    "Marca de clase": marcas_clase,
    "f": frecuencias,
})

tabla["fr"] = tabla["f"] / n
tabla["%"] = tabla["fr"] * 100
tabla["F"] = tabla["f"].cumsum()          # cumsum = suma acumulada
tabla["Fr"] = tabla["F"] / n

tabla
```

**¿Cómo se lee esta tabla?** Unos ejemplos:

- En la clase **75 - 79** hay **21 alumnos**: es la clase con más gente (la *clase modal*). Más de una cuarta parte del grupo (26.25 %) sacó entre 75 y 79.
- En la fila **70 - 74**, la frecuencia acumulada **F = 36** nos dice que **36 alumnos sacaron 74 o menos**.
- En la última fila, $F = 80$ y $F_r = 1$: claro, ahí ya juntamos a todos.

Mucho más fácil de entender que la lista de 80 números, ¿no? 😎

# Gráficas de una distribución de frecuencias

Una tabla está chida, pero una gráfica se entiende en 2 segundos. Vamos a hacer las 5 gráficas clásicas:

1. Histograma
2. Polígono de frecuencias
3. Ojivas
4. Diagrama de caja (boxplot)
5. Curva de frecuencias

## 1. Histograma

El **histograma** es una gráfica de **barras pegaditas** (sin espacio entre ellas):

- En el eje **X** van las clases (usando las **fronteras**, por eso las barras quedan pegadas: ¡aquí se ve por qué sirven los límites reales!).
- En el eje **Y** va la **frecuencia**: qué tan alta es cada barra = cuántos datos hay en esa clase.

Con matplotlib es facilísimo: `plt.hist()` recibe los datos y las fronteras (`bins`) y él solito cuenta.

```python
plt.figure(figsize=(9, 5))
plt.hist(calificaciones, bins=fronteras, edgecolor="black", color="skyblue")

plt.title("Histograma de calificaciones")
plt.xlabel("Calificaciones (fronteras de clase)")
plt.ylabel("Frecuencia (número de alumnos)")
plt.xticks(fronteras)
plt.show()
```

**¿Qué vemos?** La barra más alta está en 74.5 - 79.5 (la clase 75 - 79). Hacia los lados las barras van bajando: pocos alumnos sacaron calificaciones muy bajas o muy altas. La mayoría anda en medio, entre 60 y 90.

## 2. Polígono de frecuencias

El **polígono de frecuencias** es como "unir con líneas los techos" del histograma:

- Ponemos un punto en cada **marca de clase** (el centro de cada barra) a la altura de su **frecuencia**.
- Unimos los puntos con líneas rectas.
- **Truco importante:** agregamos una clase "fantasma" con frecuencia **0** al principio y otra al final (en 47 y en 102), para que el polígono "baje al piso" y quede cerrado.

```python
# Agregamos una marca de clase extra a cada lado, con frecuencia 0
x_poligono = [marcas_clase[0] - c] + marcas_clase + [marcas_clase[-1] + c]
y_poligono = [0] + frecuencias + [0]

plt.figure(figsize=(9, 5))
plt.plot(x_poligono, y_poligono, marker="o", color="red")

plt.title("Polígono de frecuencias")
plt.xlabel("Marcas de clase")
plt.ylabel("Frecuencia")
plt.xticks(x_poligono)
plt.grid(True, linestyle="--", alpha=0.5)
plt.show()
```

Para que se vea clarito que el polígono sale del histograma, los ponemos juntos en la misma gráfica:

```python
plt.figure(figsize=(9, 5))
plt.hist(calificaciones, bins=fronteras, edgecolor="black", color="skyblue", label="Histograma")
plt.plot(x_poligono, y_poligono, marker="o", color="red", label="Polígono de frecuencias")

plt.title("Histograma + Polígono de frecuencias")
plt.xlabel("Calificaciones")
plt.ylabel("Frecuencia")
plt.legend()
plt.show()
```

¿Ves? Cada puntito rojo está justo en medio del techo de cada barra. El polígono es útil sobre todo cuando quieres **comparar dos grupos en la misma gráfica** (por ejemplo, el grupo A contra el grupo B), porque dos histogramas encimados se ven muy revueltos, pero dos líneas se distinguen bien.

## 3. Ojivas (polígono de frecuencias acumuladas)

La **ojiva** usa la **frecuencia acumulada** en lugar de la frecuencia normal. Hay dos tipos:

### Ojiva "menor que"
Responde preguntas del tipo: **"¿cuántos alumnos sacaron menos de X?"**

- En el eje X van las **fronteras de clase** (ojo: aquí NO usamos marcas de clase, sino fronteras, porque la acumulada se cuenta "hasta el final" de cada clase).
- Empieza en 0 (en 49.5 nadie ha sacado menos) y termina en 80 (en 99.5 ya están todos). Siempre **sube**.

### Ojiva "mayor que"
Responde: **"¿cuántos alumnos sacaron más de X?"**

- Empieza en 80 (en 49.5 todos sacaron más) y termina en 0. Siempre **baja**.

```python
# Ojiva "menor que": empieza en 0 y vamos acumulando
acumulada_menor = [0] + tabla["F"].tolist()

# Ojiva "mayor que": es lo que falta para llegar a n
acumulada_mayor = [n - a for a in acumulada_menor]

print("Fronteras:       ", fronteras)
print("Acumulada menor: ", acumulada_menor)
print("Acumulada mayor: ", acumulada_mayor)
```

```python
plt.figure(figsize=(9, 5))
plt.plot(fronteras, acumulada_menor, marker="o", color="green", label='Ojiva "menor que"')
plt.plot(fronteras, acumulada_mayor, marker="s", color="purple", label='Ojiva "mayor que"')

plt.title("Ojivas")
plt.xlabel("Fronteras de clase")
plt.ylabel("Frecuencia acumulada")
plt.xticks(fronteras)
plt.grid(True, linestyle="--", alpha=0.5)
plt.legend()
plt.show()
```

**¿Cómo se lee?**

- En la ojiva verde, si buscas 69.5 en X y subes, llegas a 24: **24 alumnos sacaron menos de 69.5** (o sea, 69 o menos).
- En la morada, en 84.5 hay 17: **17 alumnos sacaron más de 84.5** (o sea, 85 o más).

🎯 **Dato curioso:** el punto donde **se cruzan las dos ojivas** es, aproximadamente, la **mediana**: el valor que deja a la mitad del grupo abajo y a la otra mitad arriba (40 y 40). Aquí se cruzan cerquita de 75. Vamos a comprobarlo:

```python
print("Mediana real de los datos:", calificaciones.median())
```

¡Sale 75! Justo donde se cruzan. 🙌

## 4. Diagrama de caja (boxplot)

El **boxplot** (diagrama de caja y bigotes) es una gráfica que resume los datos en **5 números**:

1. **Mínimo** (sin contar datos raros)
2. **Q1 – primer cuartil**: el 25 % de los alumnos sacó esto o menos.
3. **Q2 – mediana**: la mitad sacó esto o menos.
4. **Q3 – tercer cuartil**: el 75 % sacó esto o menos.
5. **Máximo** (sin contar datos raros)

Imagínate formar a los 80 alumnos del más bajo al más alto y partir la fila en 4 grupos de 20. Los cuartiles son los "cortes" entre esos grupos.

- La **caja** va de Q1 a Q3: ahí está el 50 % del centro del grupo.
- La **raya** dentro de la caja es la mediana.
- Los **bigotes** llegan hasta el dato más lejano que no sea "raro".
- Un dato es **atípico (outlier)** si se aleja mucho: si está más allá de $1.5 \times RIC$ de la caja, donde $RIC = Q3 - Q1$ (rango intercuartílico). Esos se dibujan como puntitos sueltos.

Primero calculamos los números:

```python
q1 = calificaciones.quantile(0.25)
q2 = calificaciones.quantile(0.50)
q3 = calificaciones.quantile(0.75)
ric = q3 - q1

limite_bajo = q1 - 1.5 * ric
limite_alto = q3 + 1.5 * ric

print("Q1:", q1)
print("Q2 (mediana):", q2)
print("Q3:", q3)
print("RIC (Q3 - Q1):", ric)
print("Un dato sería atípico si es menor que", limite_bajo, "o mayor que", limite_alto)
```

```python
plt.figure(figsize=(9, 3))
plt.boxplot(calificaciones, vert=False)

plt.title("Diagrama de caja de las calificaciones")
plt.xlabel("Calificaciones")
plt.grid(True, axis="x", linestyle="--", alpha=0.5)
plt.show()
```

**¿Qué nos dice?**

- La caja va de 67.75 a 82: **la mitad "del centro" del grupo** sacó entre esas calificaciones.
- La mediana (raya naranja) está casi en medio de la caja y los bigotes miden parecido de cada lado → los datos son bastante **simétricos**.
- No hay puntitos sueltos: **no hay datos atípicos**. Nadie sacó algo exageradamente bajo o alto comparado con el resto (el 53 es bajito, pero no tanto como para considerarse raro, porque el límite era 46.375).

El boxplot es buenísimo para comparar varios grupos lado a lado en una sola gráfica.

## 5. Curva de frecuencias

La **curva de frecuencias** es el polígono de frecuencias pero **suavizado**: en lugar de líneas rectas con picos, se dibuja una curva suavecita.

¿Para qué? Porque si tuviéramos muchísimos datos y clases cada vez más angostas, el polígono se iría pareciendo cada vez más a una curva. La curva nos deja ver la **forma** de los datos (¿es simétrica? ¿está cargada a un lado?) sin distraernos con los piquitos.

Lo hacemos con `make_interp_spline` de scipy, que traza una curva suave que pasa por los puntos del polígono:

```python
from scipy.interpolate import make_interp_spline

# Muchos puntos en X para que la curva se vea suave
x_suave = np.linspace(min(x_poligono), max(x_poligono), 300)

# La curva suave que pasa por los puntos del polígono
curva = make_interp_spline(x_poligono, y_poligono, k=3)
y_suave = curva(x_suave)
y_suave = np.clip(y_suave, 0, None)   # que no baje de 0 (no hay frecuencias negativas)

plt.figure(figsize=(9, 5))
plt.hist(calificaciones, bins=fronteras, edgecolor="black", color="skyblue", alpha=0.4, label="Histograma")
plt.plot(x_poligono, y_poligono, "--", color="gray", label="Polígono de frecuencias")
plt.plot(x_suave, y_suave, color="orange", linewidth=3, label="Curva de frecuencias")

plt.title("Curva de frecuencias")
plt.xlabel("Calificaciones")
plt.ylabel("Frecuencia")
plt.legend()
plt.show()
```

**Extra:** pandas también puede dibujar una curva suave directamente desde los datos con `.plot.kde()` (*estimación de densidad*). Esta curva no usa frecuencias sino **densidad** (el área bajo la curva vale 1), por eso el histograma lo dibujamos con `density=True` para que estén en la misma escala:

```python
plt.figure(figsize=(9, 5))
plt.hist(calificaciones, bins=fronteras, edgecolor="black", color="skyblue", alpha=0.4, density=True)
calificaciones.plot.kde(color="orange", linewidth=3)

plt.title("Curva de frecuencias con KDE de pandas")
plt.xlabel("Calificaciones")
plt.ylabel("Densidad")
plt.show()
```

**¿Qué forma tiene?** Parece una **campana**: alta en el centro (cerca de 75) y bajando parejito a los lados. A esta forma se le llama **simétrica** o **en forma de campana** (parecida a la famosa *distribución normal*).

Otras formas que te puedes encontrar:

- **Sesgada a la derecha** (cola larga hacia la derecha): la mayoría sacó bajo y unos poquitos muy alto. Ej: los sueldos en un país.
- **Sesgada a la izquierda** (cola larga hacia la izquierda): la mayoría sacó alto y unos poquitos muy bajo. Ej: un examen muy fácil.
- **Bimodal** (dos jorobas): hay dos grupos distintos mezclados. Ej: un grupo donde unos estudiaron y otros de plano no.

# ¿Y todo esto para qué? 🤔

Buena pregunta. Al principio parece mucho rollo para "solo contar calificaciones", pero la intención es esta:

### 1. Pasar de datos crudos a información
80 números sueltos no dicen nada. La **distribución de frecuencias** los **organiza y resume** para que un humano los entienda. Ahora podemos decir en una frase: *"la mayoría del grupo sacó entre 70 y 85, la calificación más común está entre 75 y 79, y casi nadie reprobó feo"*. Eso es lo que le sirve a un profe, a un director o a un jefe.

### 2. Entender la forma de los datos
Con el histograma, el polígono y la curva vemos **cómo se reparten** los datos: si son simétricos, si están cargados a un lado, si hay dos grupos mezclados. Esto es súper importante porque **muchas técnicas de estadística (y de machine learning) suponen que los datos tienen cierta forma**, por ejemplo, que se parecen a una campana. Si no revisas la forma primero, puedes usar la herramienta equivocada.

### 3. Detectar datos raros
El **boxplot** nos dice al instante si hay datos atípicos (outliers). En la vida real eso puede ser un error de captura (alguien escribió 900 en vez de 90), un fraude, o algo interesante que vale la pena investigar.

### 4. Responder preguntas rápido
Con las **ojivas** y las **frecuencias acumuladas** contestas cosas como *"¿cuántos sacaron menos de 70?"* o *"¿qué calificación deja al 50 % abajo?"* sin tener que contar a mano.

### 5. Entender los conceptos detrás
Las **fronteras / límites reales** existen para que no queden huecos entre clases, y las **marcas de clase** para tener un representante de cada grupo. Sin ellos, el histograma no tendría barras pegadas, el polígono no sabría dónde poner sus puntos, y más adelante no podríamos calcular la **media, la varianza o la desviación estándar de datos agrupados** (¡que usan justo las marcas de clase!).

### 6. Es el primer paso de cualquier análisis de datos
En ciencia de datos y machine learning a esto se le llama **Análisis Exploratorio de Datos (EDA)**. Antes de entrenar cualquier modelo, lo primero que hace todo el mundo es: hacer histogramas, boxplots, ver la distribución, buscar outliers... exactamente lo que hicimos aquí. Si no conoces tus datos, no puedes confiar en lo que el modelo te diga.

En resumen: **aprender a organizar y graficar datos es aprender a "escucharlos"** antes de sacar conclusiones. 📊✨
