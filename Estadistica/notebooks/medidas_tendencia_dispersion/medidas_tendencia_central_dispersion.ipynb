# Medidas de tendencia central y de dispersión 📊

En las notebooks anteriores (`DistribucionFrecuencia.ipynb` y `DistribucionFrecuencia_v2.ipynb`) organizamos las 80 calificaciones en tablas y gráficas. Eso nos dio una **imagen** de los datos.

Ahora vamos a resumir esos 80 datos en **unos poquitos números** que nos contesten dos preguntas:

2. **¿Qué tan desparramados están?** → *Medidas de dispersión*: **rango, varianza y desviación estándar**.

Piénsalo así: si alguien te pregunta *"¿qué tal le fue al grupo?"*, lo ideal es contestar algo como *"el promedio fue 75 y casi todos andan entre 65 y 85"*. Con eso ya diste el **centro** y la **dispersión**. ¡Eso es justo lo que vamos a aprender!

Vamos a usar dos tipos de ejemplos:

- **Las 80 calificaciones** del libro, para aplicarlo a datos de verdad.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

datos = [
    68, 84, 75, 82, 68, 90, 62, 88, 76, 93,
    73, 79, 88, 73, 60, 93, 71, 59, 85, 75,
    61, 65, 75, 87, 74, 62, 95, 78, 63, 72,
    66, 78, 82, 75, 94, 77, 69, 74, 68, 60,
    96, 78, 89, 61, 75, 95, 60, 79, 83, 71,
    79, 62, 67, 97, 78, 85, 76, 65, 71, 75,
    65, 80, 73, 57, 88, 78, 62, 76, 53, 74,
    86, 67, 73, 81, 72, 63, 76, 75, 85, 77
]

calificaciones = pd.Series(datos)
n = len(calificaciones)
print("Tenemos", n, "calificaciones")
```

# Parte 1: Medidas de tendencia central

Se llaman de **"tendencia central"** porque nos dicen hacia dónde **tienden** a juntarse los datos, o sea, dónde está su **centro**. Hay tres principales: la **media**, la **mediana** y la **moda**. Cada una entiende "el centro" de forma un poquito distinta.

## 1. La media (el promedio)

Es la que todos conocemos: **sumas todos los datos y divides entre cuántos son**.

$$\bar{x} = \frac{x_1 + x_2 + \dots + x_n}{n} = \frac{\sum x}{n}$$

El símbolo $\bar{x}$ se lee "x barra", y $\sum$ (sigma) significa "suma todo".

Empecemos con un ejemplo chiquito: las calificaciones de 5 amigos.

```python
amigos = [70, 75, 80, 85, 90]

suma = sum(amigos)
cuantos = len(amigos)
media_amigos = suma / cuantos

print("Suma:", suma)
print("Cuántos:", cuantos)
print("Media:", media_amigos)
```

### 🖼️ La media es el punto de equilibrio

Una forma muy bonita de entender la media: imagina que cada dato es una **pelotita del mismo peso** puesta sobre una regla. La media es el lugar exacto donde tendrías que poner el dedo para que la regla **no se caiga** hacia ningún lado. ⚖️

```python
plt.figure(figsize=(9, 2.5))
plt.scatter(amigos, [0] * len(amigos), s=300, color="skyblue", edgecolor="black", zorder=3)
plt.axhline(0, color="black")                                               # la "regla"
plt.scatter([media_amigos], [-0.15], marker="^", s=600, color="red", zorder=3)  # el punto de apoyo

plt.text(media_amigos, -0.45, f"media = {media_amigos}", ha="center", color="red", fontsize=12)
plt.title("La media es el punto de equilibrio")
plt.ylim(-0.6, 0.5)
plt.yticks([])
plt.show()
```

Ahora con las 80 calificaciones. Lo hacemos "a mano" y luego con pandas, que lo hace con `.mean()`:

```python
media_mano = sum(datos) / n
media = calificaciones.mean()

print("Media a mano:     ", media_mano)
print("Media con pandas: ", media)
```

El promedio del grupo es **75.25**.

## 2. La mediana (el de en medio)

La mediana es el valor que queda **justo a la mitad** cuando **ordenas** los datos de menor a mayor. La mitad de los datos queda abajo y la otra mitad arriba.

Es como formar a todos por estatura y fijarte quién quedó exactamente en el centro de la fila. 🧍🧍🧍

- Si hay un número **impar** de datos: es el de en medio, y ya.
- Si hay un número **par**: no hay uno solo en medio, hay dos. Entonces se **promedian esos dos**.

```python
# Ejemplo impar: 5 datos
impar = [85, 70, 90, 75, 80]
print("Datos:", impar)
print("Ordenados:", sorted(impar))
print("Mediana:", pd.Series(impar).median())
print()

# Ejemplo par: 6 datos
par = [85, 70, 90, 75, 80, 100]
print("Datos:", par)
print("Ordenados:", sorted(par))
print("Los dos de en medio son 80 y 85, su promedio es", (80 + 85) / 2)
print("Mediana:", pd.Series(par).median())
```

### 🖼️ La mediana parte los datos en dos mitades

```python
ordenados = sorted(impar)
colores = ["lightgray", "lightgray", "orange", "lightgray", "lightgray"]

plt.figure(figsize=(9, 2.5))
plt.scatter(range(1, 6), [0] * 5, s=1200, color=colores, edgecolor="black")
for i, valor in enumerate(ordenados, start=1):
    plt.text(i, 0, valor, ha="center", va="center", fontsize=12)

plt.text(1.5, -0.4, "2 datos abajo", ha="center")
plt.text(3, -0.4, "MEDIANA", ha="center", color="darkorange", fontweight="bold")
plt.text(4.5, -0.4, "2 datos arriba", ha="center")
plt.title("La mediana es el valor que queda en medio (con los datos ordenados)")
plt.xlim(0.3, 5.7)
plt.ylim(-0.6, 0.4)
plt.axis("off")
plt.show()
```

Con las 80 calificaciones (80 es par, así que se promedian los lugares 40 y 41):

```python
ordenadas = calificaciones.sort_values().tolist()
print("Lugar 40:", ordenadas[39])   # en Python se cuenta desde 0, por eso es el 39
print("Lugar 41:", ordenadas[40])

mediana = calificaciones.median()
print("Mediana:", mediana)
```

La mediana es **75**: la mitad del grupo sacó 75 o menos, y la otra mitad 75 o más.

## 3. La moda (el más popular)

La moda es el valor que **más se repite**. Así de fácil. Es como la canción "de moda": la que más veces suena. 🎵

- Puede haber **una** moda (unimodal).
- Puede haber **dos** (bimodal) o más, si hay empate.
- O puede **no haber** moda, si ningún valor se repite.

Pandas la saca con `.mode()`. Ojo: regresa una lista, porque puede haber varias modas.

```python
print("Ejemplo con una moda:", pd.Series([3, 5, 5, 7, 8]).mode().tolist())
print("Ejemplo con dos modas:", pd.Series([3, 3, 5, 7, 7]).mode().tolist())
print()

moda = calificaciones.mode()[0]
print("Moda de las 80 calificaciones:", moda)
print("¿Cuántas veces aparece?", (calificaciones == moda).sum())
```

### 🖼️ La moda es la barra más alta

Si contamos cuántas veces aparece cada calificación y lo graficamos, la moda salta a la vista:

```python
conteo = calificaciones.value_counts().sort_index()
colores = ["orange" if valor == moda else "skyblue" for valor in conteo.index]

plt.figure(figsize=(12, 4))
plt.bar(conteo.index, conteo.values, color=colores, edgecolor="black")
plt.title(f"¿Cuántas veces aparece cada calificación? La moda es {moda}")
plt.xlabel("Calificación")
plt.ylabel("Veces que aparece")
plt.show()
```

El 75 aparece **7 veces**, más que cualquier otra calificación. 🏆

> 💡 La moda es la **única** medida que sirve para datos que no son números. Por ejemplo, si preguntas el color favorito de tu grupo, no puedes sacar el "promedio de colores", pero sí la moda: "el color más votado fue el azul".

## 🖼️ Las tres juntas

Pongamos media, mediana y moda sobre el histograma de las calificaciones (el mismo de la notebook 1):

```python
fronteras = np.arange(49.5, 100, 5)

plt.figure(figsize=(10, 5))
plt.hist(calificaciones, bins=fronteras, edgecolor="black", color="lightgray")
plt.axvline(media, color="red", linewidth=3, label=f"Media = {media}")
plt.axvline(mediana, color="orange", linewidth=3, linestyle="--", label=f"Mediana = {mediana}")
plt.axvline(moda, color="green", linewidth=3, linestyle=":", label=f"Moda = {moda}")

plt.title("Media, mediana y moda de las calificaciones")
plt.xlabel("Calificaciones")
plt.ylabel("Frecuencia")
plt.legend()
plt.show()
```

¡Las tres casi encimadas! Media 75.25, mediana 75 y moda 75. Cuando eso pasa, es señal de que los datos son **simétricos**: tienen forma de campana, con el centro bien definido. Eso coincide con lo que vimos en la curva de frecuencias de la notebook 1. 👌

## ¿Y cuál uso? El problema de los datos extremos 😱

La media tiene un punto débil: **se deja jalar por los valores extremos**.

Ejemplo: en una oficinita trabajan 5 empleados y el jefe. Estos son sus sueldos mensuales:

```python
sueldos = pd.Series([9000, 10000, 10000, 11000, 12000, 150000])   # el último es el jefe

print("Media de sueldos:  ", round(sueldos.mean(), 2))
print("Mediana de sueldos:", sueldos.median())
```

¡La media dice que en promedio ganan **33,667**! Pero **nadie** gana eso: 5 de 6 ganan entre 9,000 y 12,000. El sueldo del jefe "jaló" el promedio hacia arriba.

La mediana, en cambio, dice **10,500**, que sí representa bien a la mayoría. A la mediana **no le importa qué tan grande sea el extremo**: si el jefe ganara un millón, la mediana seguiría siendo 10,500.

### 🖼️ Así se ve:

```python
plt.figure(figsize=(10, 2.8))
plt.scatter(sueldos, [0] * len(sueldos), s=200, color="skyblue", edgecolor="black", zorder=3)
plt.axvline(sueldos.mean(), color="red", linewidth=3, label=f"Media = {sueldos.mean():,.0f}")
plt.axvline(sueldos.median(), color="orange", linewidth=3, linestyle="--", label=f"Mediana = {sueldos.median():,.0f}")

plt.text(150000, 0.02, "el jefe", ha="center", va="bottom")
plt.title("Un dato extremo jala a la media, pero no a la mediana")
plt.xlabel("Sueldo mensual")
plt.yticks([])
plt.legend(loc="center")
plt.show()
```

Por eso, cuando escuches noticias como "el ingreso promedio de un país", desconfía un poquito: unos cuantos millonarios pueden hacer que el promedio se vea mucho más alto de lo que gana la mayoría. 😉

**Regla práctica:**

| Situación | Mejor medida |
|---|---|
| Datos simétricos, sin valores extremos | **Media** |
| Hay valores extremos o datos sesgados (sueldos, precios de casas) | **Mediana** |
| Datos que no son números (colores, marcas, carreras) | **Moda** |

### 🖼️ Dónde quedan según la forma de los datos

Cuando los datos **no** son simétricos, la media se va hacia el lado de la "cola" (donde están los extremos), y la mediana se queda más cerca del montón. Vamos a simular tres grupos de datos para verlo (no te preocupes por cómo se generan, solo fíjate en las líneas):

```python
generador = np.random.default_rng(0)       # para que siempre salgan los mismos números
simetricos = generador.normal(70, 10, 1000)
sesgo_derecha = generador.exponential(10, 1000) + 50
sesgo_izquierda = 100 - generador.exponential(10, 1000)

ejemplos = [
    ("Simétrica", simetricos),
    ("Sesgada a la derecha", sesgo_derecha),
    ("Sesgada a la izquierda", sesgo_izquierda),
]

fig, ejes = plt.subplots(1, 3, figsize=(15, 4))
for eje, (titulo, valores) in zip(ejes, ejemplos):
    eje.hist(valores, bins=30, color="lightgray", edgecolor="black")
    eje.axvline(np.mean(valores), color="red", linewidth=3, label="Media")
    eje.axvline(np.median(valores), color="orange", linewidth=3, linestyle="--", label="Mediana")
    eje.set_title(titulo)
    eje.legend()

plt.show()
```

- **Simétrica:** media y mediana encimadas (como nuestras calificaciones).
- **Sesgada a la derecha** (cola hacia la derecha, como los sueldos): la media se va a la **derecha** de la mediana.
- **Sesgada a la izquierda** (cola hacia la izquierda, como un examen muy fácil): la media se va a la **izquierda**.

> 🧠 Truquito para acordarte: **la media siempre persigue a la cola**.

## Bonus: la media de datos agrupados

¿Te acuerdas de las **marcas de clase** de la notebook 1? Aquí es donde sirven. Si solo tuvieras la tabla de frecuencias (y no los 80 datos originales), puedes **estimar** la media haciendo de cuenta que todos los datos de cada clase valen su marca de clase:

$$\bar{x} \approx \frac{\sum (f \cdot X_m)}{n}$$

```python
marcas_clase = [52, 57, 62, 67, 72, 77, 82, 87, 92, 97]
frecuencias = [1, 2, 11, 10, 12, 21, 6, 9, 4, 4]       # de la tabla de la notebook 1

suma_f_por_xm = sum(f * xm for f, xm in zip(frecuencias, marcas_clase))
media_agrupada = suma_f_por_xm / sum(frecuencias)

print("Media con datos agrupados:", media_agrupada)
print("Media real:               ", media)
```

¡75.375 contra 75.25! Casi igual. No es exacta porque al agrupar perdemos un poquito de información, pero se acerca muchísimo.

# Parte 2: Medidas de dispersión

Ya sabemos dónde está el **centro**. Pero eso no basta. Mira estos dos grupos de 5 alumnos:

```python
grupo_a = pd.Series([78, 79, 80, 81, 82])
grupo_b = pd.Series([60, 70, 80, 90, 100])

print("Media grupo A:", grupo_a.mean(), " | Mediana:", grupo_a.median())
print("Media grupo B:", grupo_b.mean(), " | Mediana:", grupo_b.median())
```

¡Los dos tienen media 80 y mediana 80! Si solo te dijera eso, pensarías que son iguales. Pero mira cómo se ven:

### 🖼️ Mismo centro, diferente dispersión

```python
plt.figure(figsize=(10, 3))
plt.scatter(grupo_a, [1] * 5, s=200, color="green", edgecolor="black", label="Grupo A")
plt.scatter(grupo_b, [0] * 5, s=200, color="purple", edgecolor="black", label="Grupo B")
plt.axvline(80, color="red", linestyle="--", label="Media = 80 (ambos)")

plt.yticks([0, 1], ["Grupo B", "Grupo A"])
plt.ylim(-0.6, 1.6)
plt.title("Los dos grupos tienen la misma media, pero ¡qué diferentes son!")
plt.xlabel("Calificación")
plt.legend(loc="upper left")
plt.show()
```

- En el **grupo A** todos sacaron casi lo mismo: son **parejitos**, el promedio los representa súper bien.
- En el **grupo B** hay de todo: uno sacó 60 y otro 100. Están **desparramados**. Decir "el promedio es 80" esconde que hubo alguien que casi reprueba.

Las **medidas de dispersión** sirven para medir justo eso: **qué tan separados están los datos**.

## 1. El rango

La más sencilla: **mayor menos menor**. Ya la usamos en la notebook 1.

```python
print("Rango grupo A:", grupo_a.max() - grupo_a.min())
print("Rango grupo B:", grupo_b.max() - grupo_b.min())
print("Rango de las 80 calificaciones:", calificaciones.max() - calificaciones.min())
```

El rango es fácil, pero tiene un problema: **solo usa dos datos** (el mayor y el menor) e ignora a todos los demás. Si hay un solo dato extremo, el rango se dispara. Necesitamos algo que tome en cuenta a **todos** los datos.

## 2. Las desviaciones: ¿qué tan lejos está cada dato de la media?

La idea es: para cada dato, calculamos **qué tan lejos está de la media**. A eso se le llama **desviación**:

$$\text{desviación} = x - \bar{x}$$

- Si es **positiva**, el dato está **arriba** de la media.
- Si es **negativa**, está **abajo**.

```python
media_b = grupo_b.mean()
desviaciones = grupo_b - media_b

print("Datos:       ", grupo_b.tolist())
print("Media:       ", media_b)
print("Desviaciones:", desviaciones.tolist())
```

### 🖼️ Las desviaciones, dibujadas

Cada barra es la distancia de un alumno a la media (la línea roja en 0):

```python
colores = ["tomato" if d < 0 else "mediumseagreen" for d in desviaciones]

plt.figure(figsize=(8, 4))
plt.bar(["Alumno 1", "Alumno 2", "Alumno 3", "Alumno 4", "Alumno 5"], desviaciones, color=colores, edgecolor="black")
plt.axhline(0, color="red", linewidth=2)

plt.title("Desviaciones del grupo B respecto a su media (80)")
plt.ylabel("Distancia a la media")
plt.show()
```

Ahora, para tener **una sola medida**, se nos podría ocurrir promediar las desviaciones... pero mira lo que pasa si las sumamos:

```python
print("Suma de las desviaciones:", desviaciones.sum())
```

¡**Cero**! 😵 Y esto **siempre** pasa, con cualquier conjunto de datos. ¿Por qué? Por lo del punto de equilibrio: la media está justo donde lo de arriba compensa a lo de abajo. Las barras verdes y rojas se cancelan.

**Solución:** elevar cada desviación **al cuadrado**. Así todas quedan positivas (porque negativo por negativo da positivo) y ya no se cancelan.

## 3. La varianza

La **varianza** es el **promedio de las desviaciones al cuadrado**:

$$s^2 = \frac{\sum (x - \bar{x})^2}{n - 1}$$

Paso a paso:
1. Sacas la media.
2. A cada dato le restas la media (desviación).
3. Elevas al cuadrado cada desviación.
4. Las sumas.
5. Divides entre $n - 1$. (Ahorita te explico por qué $n - 1$ y no $n$. 😉)

```python
cuadrados = desviaciones ** 2

tabla_b = pd.DataFrame({
    "x": grupo_b,
    "x - media": desviaciones,
    "(x - media)²": cuadrados,
})
print(tabla_b)
print()

suma_cuadrados = cuadrados.sum()
varianza_b = suma_cuadrados / (len(grupo_b) - 1)

print("Suma de cuadrados:", suma_cuadrados)
print("Varianza (a mano):", varianza_b)
print("Varianza (pandas):", grupo_b.var())
```

### 🖼️ ¿Por qué "al cuadrado"? Literalmente son cuadrados

Elevar al cuadrado una desviación es como dibujar un **cuadrado** cuyo lado mide esa desviación. La varianza es, más o menos, el **tamaño promedio** de esos cuadrados:

```python
fig, ejes = plt.subplots(1, 2, figsize=(12, 5), sharey=True)

for eje, grupo, nombre, color in [(ejes[0], grupo_a, "Grupo A", "green"), (ejes[1], grupo_b, "Grupo B", "purple")]:
    media_g = grupo.mean()
    for x in grupo:
        lado = abs(x - media_g)
        inicio = min(x, media_g)
        eje.add_patch(plt.Rectangle((inicio, 0), lado, lado, facecolor=color, alpha=0.3, edgecolor="black", linewidth=2))
    eje.scatter(grupo, [0] * 5, s=120, color=color, edgecolor="black", zorder=3)
    eje.axvline(media_g, color="red", linestyle="--")
    eje.set_title(f"{nombre}: varianza = {grupo.var()}")
    eje.set_xlim(55, 105)
    eje.set_aspect("equal")
    eje.set_xlabel("Calificación")

plt.suptitle("Cada cuadrado tiene como lado la distancia del dato a la media")
plt.show()
```

En el grupo A los cuadrados son **diminutos** (casi ni se ven), porque todos están pegados a la media. En el grupo B son **enormes**. ¡Por eso su varianza es 250 contra 2.5!

**El problema de la varianza:** como elevamos al cuadrado, también se elevan al cuadrado las **unidades**. Si los datos son calificaciones, la varianza está en "calificaciones al cuadrado" 🤨, lo cual no tiene mucho sentido. Ahí entra la desviación estándar.

## 4. La desviación estándar

Es simplemente la **raíz cuadrada de la varianza**:

$$s = \sqrt{s^2}$$

Con la raíz "deshacemos" el cuadrado y regresamos a las **unidades originales** (calificaciones). Se puede interpretar como **qué tan lejos de la media está un dato "típico"**, en promedio.

```python
print("Desviación estándar grupo A:", round(grupo_a.std(), 2))
print("Desviación estándar grupo B:", round(grupo_b.std(), 2))
print()
print("Comprobación (raíz de la varianza del B):", round(varianza_b ** 0.5, 2))
```

- **Grupo A:** desviación de **1.58** → los alumnos están, en promedio, a poco más de 1 punto del 80. Súper parejitos.
- **Grupo B:** desviación de **15.81** → están, en promedio, a casi 16 puntos del 80. ¡Muy disparejos!

Ahora sí, **un solo número** nos dice qué tan desparramados están. 🙌

## ¿Por qué $n - 1$? (Población vs. muestra)

Hay dos fórmulas, y la diferencia es entre qué número divides:

| | Se usa cuando... | Divides entre | Símbolos |
|---|---|---|---|
| **Población** | Tienes **todos** los datos que te interesan | $n$ | $\sigma^2$, $\sigma$ (sigma) |
| **Muestra** | Tienes **una parte** y quieres sacar conclusiones del todo | $n - 1$ | $s^2$, $s$ |

¿Por qué $n - 1$ en la muestra? Explicación corta y relajada: cuando solo tienes una muestra, los datos tienden a estar **más cerca de su propia media** que de la media real de toda la población, así que la dispersión sale un poquito **más chica** de lo que es en realidad. Dividir entre $n - 1$ (un número más chico) hace el resultado un poquito **más grande** y **corrige** ese error.

⚠️ **¡Ojo con Python!** Cada librería usa una por default:
- **pandas** `.var()` y `.std()` → **muestra** ($n - 1$).
- **numpy** `np.var()` y `np.std()` → **población** ($n$).

Puedes cambiarlo con el parámetro `ddof` ("delta de grados de libertad"): `ddof=1` es muestra, `ddof=0` es población.

```python
print("pandas .std() (muestra, n-1):        ", round(grupo_b.std(), 4))
print("pandas .std(ddof=0) (población, n):  ", round(grupo_b.std(ddof=0), 4))
print("numpy np.std() (población, n):       ", round(np.std(grupo_b), 4))
print("numpy np.std(ddof=1) (muestra, n-1): ", round(np.std(grupo_b, ddof=1), 4))
```

Con 5 datos la diferencia se nota (15.81 vs 14.14). Con muchos datos casi no se nota, porque dividir entre 80 o entre 79 da casi lo mismo.

## Ahora con las 80 calificaciones

```python
varianza = calificaciones.var()
desviacion = calificaciones.std()

print("Varianza (muestra):           ", round(varianza, 2))
print("Desviación estándar (muestra):", round(desviacion, 2))
print("Desviación estándar (población):", round(calificaciones.std(ddof=0), 2))
```

La desviación estándar es **10.37**. O sea: en promedio, cada alumno está a unos **10 puntos** del promedio del grupo (75.25).

### 🖼️ La desviación estándar sobre el histograma

Vamos a sombrear las zonas que quedan a **1** y a **2** desviaciones estándar de la media:

```python
plt.figure(figsize=(10, 5))
plt.axvspan(media - 2 * desviacion, media + 2 * desviacion, color="gold", alpha=0.25, label="Media ± 2 desv. estándar")
plt.axvspan(media - desviacion, media + desviacion, color="orange", alpha=0.35, label="Media ± 1 desv. estándar")
plt.hist(calificaciones, bins=fronteras, edgecolor="black", color="skyblue")
plt.axvline(media, color="red", linewidth=3, label=f"Media = {media}")

plt.title("¿Qué tan lejos de la media están las calificaciones?")
plt.xlabel("Calificaciones")
plt.ylabel("Frecuencia")
plt.legend(loc="upper left")
plt.show()
```

### La regla empírica (68 - 95 - 99.7)

Cuando los datos tienen forma de **campana** (como los nuestros), se cumple más o menos esta regla mágica:

- Cerca del **68 %** de los datos está a **1** desviación estándar de la media.
- Cerca del **95 %** está a **2** desviaciones estándar.
- Casi el **99.7 %** (prácticamente todos) está a **3** desviaciones estándar.

¿Se cumplirá en nuestras calificaciones? ¡Vamos a contar!

```python
for k in [1, 2, 3]:
    abajo = media - k * desviacion
    arriba = media + k * desviacion
    cuantos = ((calificaciones >= abajo) & (calificaciones <= arriba)).sum()
    porcentaje = cuantos / n * 100
    print(f"Entre {abajo:.1f} y {arriba:.1f} ({k} desv.): {cuantos} alumnos = {porcentaje:.1f} %")
```

**65 %, 96 % y 100 %.** ¡Muy cerquita de 68 - 95 - 99.7! 🎯 Eso confirma que nuestras calificaciones tienen una forma bastante parecida a la campana (la *distribución normal*).

Esto es súper útil: con solo saber la **media (75.25)** y la **desviación estándar (10.37)**, sin ver los datos, ya puedes decir: *"la mayoría del grupo (2 de cada 3) sacó entre 65 y 86, y casi todos entre 54 y 96"*.

## Bonus: el coeficiente de variación

A veces quieres comparar la dispersión de cosas que están en **escalas distintas** (por ejemplo, estaturas en cm contra pesos en kg). La desviación estándar no sirve para eso directamente, pero el **coeficiente de variación** sí: es la desviación estándar **como porcentaje de la media**.

$$CV = \frac{s}{\bar{x}} \times 100$$

```python
cv = desviacion / media * 100
print(f"Coeficiente de variación de las calificaciones: {cv:.1f} %")
```

Un 13.8 % se considera una dispersión **baja/moderada**: el grupo es relativamente homogéneo. (Como guía informal: menos de 15 % es poca dispersión, más de 30 % es mucha.)

# ¿Y todo esto para qué? 🤔

### 1. Resumir muchísimos datos en poquitos números
Nadie quiere leer 80 calificaciones (y mucho menos 80 millones de datos, que es lo normal en ciencia de datos). Con **un número para el centro** y **otro para la dispersión** ya entiendes lo esencial: *"el promedio fue 75 con una desviación de 10"*.

### 2. El centro solo cuenta la mitad de la historia
Los grupos A y B tenían la misma media, pero eran completamente distintos. **Siempre que des un promedio, acompáñalo de una medida de dispersión.** Un promedio solo puede engañar.

### 3. Elegir la medida correcta evita conclusiones falsas
El ejemplo de los sueldos lo deja clarito: si hay valores extremos, la media miente y la mediana dice la verdad. Saber cuál usar te protege de malinterpretar datos (y de que te engañen con ellos en noticias o publicidad 📺).

### 4. Comparar grupos de forma justa
¿Qué grupo fue más parejo? ¿Qué máquina de una fábrica es más precisa? ¿Qué inversión es más riesgosa? Todo eso se contesta comparando **medias y desviaciones estándar**. En finanzas, por ejemplo, la desviación estándar se usa directamente como medida del **riesgo**.

### 5. Saber qué es "normal" y qué es "raro"
Con la media y la desviación estándar sabes si un dato es común o extraño. Un alumno con 97 está a unas 2 desviaciones arriba de la media: es de los mejores del grupo. Esta idea es la base para **detectar fraudes, fallas o datos atípicos**.

### 6. Es la base de todo lo que sigue
Estos conceptos aparecen **en todos lados** más adelante:
- La **distribución normal** y la regla 68-95-99.7 → probabilidad.
- Los **intervalos de confianza** y las **pruebas de hipótesis** → usan la media y la desviación estándar.
- En **machine learning**, antes de entrenar un modelo casi siempre se **estandarizan** los datos: a cada dato se le resta la media y se divide entre la desviación estándar ($z = \frac{x - \bar{x}}{s}$). Muchos algoritmos funcionan mucho mejor así. Y la **varianza** es la idea central de técnicas como PCA o de cómo se mide el error de un modelo.

En resumen: la **tendencia central** te dice **dónde** están los datos, y la **dispersión** te dice **qué tan confiable** es ese centro. Juntas son la primera foto que le tomas a cualquier conjunto de datos. 📸✨
